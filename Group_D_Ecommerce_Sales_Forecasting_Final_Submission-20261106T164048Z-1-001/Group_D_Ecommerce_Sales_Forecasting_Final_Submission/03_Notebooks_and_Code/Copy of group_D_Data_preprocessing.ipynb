```python
# Create a copy of the original dataset
df_clean = df.copy()

# Remove exact duplicate rows
df_clean = df_clean.drop_duplicates().copy()

print("Original dataset:", df.shape)
print("Cleaned dataset:", df_clean.shape)
print("Duplicates remaining:", df_clean.duplicated().sum())
```

```python
df_clean['Revenue'] = df_clean['Quantity'] * df_clean['UnitPrice']

# Identify bad debt adjustment records
df_clean['Is_Bad_Debt_Adjustment'] = (
    df_clean['Description']
    .str.contains('adjust bad debt', case=False, na=False)
)

print("Bad debt adjustment records:",
      df_clean['Is_Bad_Debt_Adjustment'].sum())

display(df_clean[df_clean['Is_Bad_Debt_Adjustment']])
```

```python
# Exclude bad debt adjustments from sales revenue calculations
sales_transactions = df_clean[
    ~df_clean['Is_Bad_Debt_Adjustment']
].copy()

print("Sales transaction records:", len(sales_transactions))
print("Total sales revenue:", sales_transactions['Revenue'].sum())
```

```python
# Extract the date from InvoiceDate
sales_transactions['Date'] = (
    sales_transactions['InvoiceDate'].dt.date
)

# Calculate daily net revenue
daily_net_revenue = (
    sales_transactions.groupby('Date')['Revenue']
    .sum()
    .reset_index()
)

# Rename the columns
daily_net_revenue.columns = ['Date', 'NetRevenue']

# Sort by date
daily_net_revenue = daily_net_revenue.sort_values('Date')

# Display the first 10 days
display(daily_net_revenue.head(10))

print("Number of days:", len(daily_net_revenue))
print("Total net revenue:", daily_net_revenue['NetRevenue'].sum())
```

```python
# Convert Date column to datetime
daily_net_revenue['Date'] = pd.to_datetime(
    daily_net_revenue['Date']
)

# Create a complete daily date range
all_dates = pd.date_range(
    start=daily_net_revenue['Date'].min(),
    end=daily_net_revenue['Date'].max(),
    freq='D'
)

# Include dates with no recorded transactions
daily_net_revenue = (
    daily_net_revenue
    .set_index('Date')
    .reindex(all_dates, fill_value=0)
    .rename_axis('Date')
    .reset_index()
)

# Display results
display(daily_net_revenue.head(15))

print("Total calendar days:", len(daily_net_revenue))
print("Total net revenue:", daily_net_revenue['NetRevenue'].sum())
print("Missing values:", daily_net_revenue.isnull().sum())
```

```python
# Save cleaned transactions, including flagged adjustments
df_clean.to_csv('cleaned_online_retail.csv', index=False)

# Save daily net revenue
daily_net_revenue.to_csv('daily_net_revenue.csv', index=False)

# Save sales transactions excluding bad debt adjustments
sales_transactions.to_csv('sales_transactions.csv', index=False)

print("All three files have been saved successfully!")
```

Preprocessing decisions
- Exact duplicate transaction rows were removed.
- Line revenue was calculated as Quantity × UnitPrice.
- Transactions with negative quantities or negative revenue were retained so returns and cancellations reduce net revenue.
- Three rows identified as bad-debt adjustments were flagged and excluded from sales transactions and daily net revenue.
- Dates with no recorded transactions were included in the daily series with revenue set to zero.
- Missing customer IDs and descriptions were retained; they may limit customer-level or product-level analysis, but they do not prevent daily revenue aggregation.
